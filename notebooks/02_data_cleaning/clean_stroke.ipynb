# DISEASE: Stroke

### 1. Load

```python
import pandas as pd
from clean_datasets import age_years_to_ranges, normalize_categorical, convert_to_binary

df_raw = pd.read_csv("C:/Users/User/Documents/diploma/data/datasets_original/healthcare-dataset-stroke-data.csv")
df_raw = df_raw[df_raw["age"] >= 18].copy()

print(df_raw.shape)
df_raw.head()
```

### 2. Rename columns

```python
df = pd.DataFrame(index=df_raw.index)

df["heart_disease"] = df_raw["heart_disease"].astype("Int8")
df["hypertension"] = df_raw["hypertension"].astype("Int8")
df["stroke"] = df_raw["stroke"].astype("Int8")

gender_norm = normalize_categorical(df_raw["gender"])

df["sex_binary"] = gender_norm.map({
    "female": 0,
    "male": 1
}).astype("Int8")

df["sex"] = gender_norm

df["age"] = df_raw["age"].astype("float")
df["age_range"] = age_years_to_ranges(df_raw["age"])

df["bmi"] = df_raw["bmi"].astype("float")

smoke_norm = normalize_categorical(df_raw["smoking_status"])
df["smoking_status"] = smoke_norm

df["glucose"] = df_raw["avg_glucose_level"].astype("float")

df["ever_married"] = convert_to_binary(df_raw["ever_married"])
df["work_type"] = normalize_categorical(df_raw["work_type"])

residence_norm = normalize_categorical(df_raw["Residence_type"])
df["residence_type"] = residence_norm.map({
    "urban": 1,
    "rural": 0
}).astype("Int8")

df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation matrix: Stroke dataset")
plt.tight_layout()
plt.show()
```

```python
target = "stroke"
corr[target].drop(target).sort_values(key=abs, ascending=False)
```

### 3. Save cleaned dataset

```python
df.to_csv("C:/Users/User/Documents/diploma/data/datasets_cleaned/stroke_clean.csv", index=False)

print(f"Saved {len(df)} rows × {df.shape[1]} columns")
```
